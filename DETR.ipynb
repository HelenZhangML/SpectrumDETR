**Out-of-Distribution Generalisation Performance**

```python
import importlib
import scripts.plot_ood_pca as ood_pca

importlib.reload(ood_pca)

from scripts.plot_ood_pca import (
    get_combo_paths,
    load_env_matrix_from_raw_pt,
    make_loader,
    load_model_from_ckpt,
    extract_pred_env_vectors,
    build_training_subset_masks,
    plot_combo_with_subsets,
)

# --------------------------------------------------
# CONFIG
# --------------------------------------------------
REPS = ["skipatom", "matscholar", "soap"]
HOLDOUTS = ["CBr", "carbonyl_O", "aromatic_C"]

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
BATCH_SIZE = 8
CONF_THRESH = 0.4
MAX_TRAIN_POINTS_FOR_PLOT = 5000

OUT_DIR = Path("ood_pca_plots_with_subsets")
OUT_DIR.mkdir(exist_ok=True)

HOLDOUT_LABELS = {
    "CBr": "C–Br",
    "carbonyl_O": "carbonyl O",
    "aromatic_C": "aromatic C",
}

SUBSET_COLORS = {
    "C–Cl": "blue",
    "C–F": "purple",
    "other O": "blue",
    "other C": "blue",
}

summary = []

for rep in REPS:
    for holdout in HOLDOUTS:
        print("\n" + "=" * 100)
        print(f"{rep} | {holdout}")
        print("=" * 100)

        paths = get_combo_paths(rep, holdout)
        for k, v in paths.items():
            print(f"{k:9s}: {v}")

        required = ["train_pt", "ood_pt", "ckpt", "train_csv"]
        if any(paths[k] is None for k in required):
            print("Skipping because required files are missing.")
            summary.append({
                "rep": rep,
                "holdout": holdout,
                "status": "missing_files",
                **paths,
            })
            continue

        try:
            X_train_128 = load_env_matrix_from_raw_pt(paths["train_pt"])
            X_ood_true_128 = load_env_matrix_from_raw_pt(paths["ood_pt"])

            ood_loader = make_loader(paths["ood_pt"], batch_size=BATCH_SIZE)
            model = load_model_from_ckpt(paths["ckpt"], device=DEVICE)

            X_pred_128 = extract_pred_env_vectors(
                model,
                ood_loader,
                DEVICE,
                conf_thresh=CONF_THRESH,
            )

            print("Train reference envs:", X_train_128.shape)
            print("True OOD envs:      ", X_ood_true_128.shape)
            print("Predicted envs:     ", X_pred_128.shape)

            if X_ood_true_128.shape[0] == 0:
                raise ValueError("True OOD env matrix is empty.")

            if X_pred_128.shape[0] == 0:
                raise ValueError("Predicted env matrix is empty. Lower CONF_THRESH to 0.4 or 0.3.")

            merged_df, subset_masks = build_training_subset_masks(paths["train_csv"], holdout)
            for label, mask in subset_masks.items():
                print(f"{label} count: {int(mask.sum())}")

            plot_combo_with_subsets(
                rep=rep,
                holdout=holdout,
                X_train_128=X_train_128,
                X_ood_true_128=X_ood_true_128,
                X_pred_128=X_pred_128,
                train_csv_path=paths["train_csv"],
                out_dir=OUT_DIR,
                show_saved_image=False,
            )

            summary.append({
                "rep": rep,
                "holdout": holdout,
                "status": "ok",
                "n_train": int(X_train_128.shape[0]),
                "n_ood_true": int(X_ood_true_128.shape[0]),
                "n_pred": int(X_pred_128.shape[0]),
            })

        except Exception as e:
            print(f"FAILED: {type(e).__name__}: {e}")
            summary.append({
                "rep": rep,
                "holdout": holdout,
                "status": "failed",
                "error": f"{type(e).__name__}: {e}",
                **paths,
            })

print("\nSUMMARY")
for row in summary:
    print(row)

print("\nSaved plots in:", OUT_DIR.resolve())
```

Recovery of environment types across unseen polymers plot

```python
from IPython.display import display

from scripts.evaluate_unseen_polymers import run_all_cases


# Font sizes 
plt.rcParams.update({
    "font.size": 14,
    "axes.labelsize": 15,
    "xtick.labelsize": 13,
    "ytick.labelsize": 13,
    "legend.fontsize": 13,
    "legend.title_fontsize": 14,
    "figure.titlesize": 18,
})

# ============================================================
# RUN UNSEEN-POLYMER EVALUATION
# ============================================================
outputs, all_summary = run_all_cases(
    polymers=["pDMA Sample 1", "pHEMA Sample 2", "pHPMA Sample 3"],
    representations=["SkipAtom", "MatScholar", "SOAP"],
    regimes=["full", "holdout"],
    use_corrected=False,
    conf_thresh=0.30,
    show_plots=False,
)

all_summary.to_csv("unseen_vgd_groundtruth_comparison_summary.csv", index=False)
display(all_summary)

# ============================================================
# ENVIRONMENT-LEVEL RECOVERY FRACTION
# ============================================================
ENV_ORDER = ["aliphatic C", "C-O", "other O", "carbonyl C", "carbonyl O"]
REP_ORDER = ["SkipAtom", "MatScholar", "SOAP"]
REGIME_ORDER = ["full", "holdout"]

rows = []

for out in outputs:
    polymer = out["polymer"]
    rep = out["representation"]
    regime = out["regime"]
    comp = out["comparison"].copy()

    for env in ENV_ORDER:
        sub = comp[comp["environment_label"] == env]

        true_present = len(sub) > 0 and int(sub["true_count"].iloc[0]) > 0
        pred_present = len(sub) > 0 and int(sub["predicted_count"].iloc[0]) > 0

        if true_present:
            rows.append({
                "polymer": polymer,
                "representation": rep,
                "regime": regime,
                "environment_label": env,
                "recovered": int(pred_present),
            })

recovery_long = pd.DataFrame(rows)

recovery_summary = (
    recovery_long
    .groupby(["regime", "representation", "environment_label"], as_index=False)
    .agg(
        recovery_fraction=("recovered", "mean"),
        n_polymers=("polymer", "nunique"),
    )
)

recovery_summary.to_csv("unseen_environment_recovery_fraction.csv", index=False)
display(recovery_summary)

# ============================================================
# PLOT REPORT FIGURE
# ============================================================
colors = {
    "SkipAtom": "0.6",
    "MatScholar": "0.3",
    "SOAP": "seagreen",
}

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

for ax, regime in zip(axes, REGIME_ORDER):
    sub = recovery_summary[recovery_summary["regime"] == regime].copy()

    x = np.arange(len(ENV_ORDER))
    width = 0.24

    for i, rep in enumerate(REP_ORDER):
        vals = []

        for env in ENV_ORDER:
            row = sub[
                (sub["representation"] == rep)
                & (sub["environment_label"] == env)
            ]

            if len(row):
                vals.append(float(row["recovery_fraction"].iloc[0]))
            else:
                vals.append(0.0)

        ax.bar(
            x + (i - 1) * width,
            vals,
            width=width,
            label=rep,
            color=colors[rep],
            edgecolor="black",
            linewidth=0.5,
        )

    xtick_labels = []

    for env in ENV_ORDER:
        n = recovery_long[
            (recovery_long["regime"] == regime)
            & (recovery_long["environment_label"] == env)
        ]["polymer"].nunique()

        xtick_labels.append(f"{env}\n(n={n})")

    ax.set_title("Full" if regime == "full" else "Holdout", pad=10)
    ax.set_xticks(x)
    ax.set_xticklabels(xtick_labels, rotation=40, ha="right")
    ax.set_ylim(0, 1.05)

    ax.set_ylabel("Recovery fraction")

axes[1].legend(
    title="Representation",
    frameon=False,
    loc="upper right",
)

plt.tight_layout()

plt.savefig(
    "recoveryfraction.png",
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print("Saved: recoveryfraction.png")
print("Saved: unseen_environment_recovery_fraction.csv")
```
